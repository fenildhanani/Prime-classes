```python
import seaborn as sns
from sklearn.datasets import load_iris
import pandas as pd
from sklearn.cluster import KMeans
```

```python
iris =load_iris()
```

```python
X =iris.data
# X = pd.DataFrame(iris.data)
# X.columns = iris.feature_names
y = iris.target
```

```python
#visulize
sns.scatterplot(x=X[:, 0],y=X[:, 2],c=y)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
# x_scaled
```

```python
# optional Dimentionality reduction using PCA
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
pca_data =pca.fit_transform(X)
```

```python
# elbow method
wcss = []

for k in range(1,11):
    kmeans = KMeans(n_clusters=k)
    kmeans.fit_predict(pca_data)
    wcss.append(kmeans.inertia_)


    
```

```python
wcss
```

```python
sns.lineplot(x=range(1,11), y =wcss,marker = 'o')
```

```python
kmeans = KMeans(n_clusters=3)
lables = kmeans.fit_predict(X_scaled)

sns.scatterplot(x=X[:, 0],y=X[:, 2])
```

```python
# kmeans
kmeans = KMeans(n_clusters=3,random_state=10)
labels = kmeans.fit_predict(pca_data)

sns.scatterplot(x=X[:,0],y=pca_data[:, 1] , c=labels)
sns.scatterplot(x=kmeans.cluster_centers_[:, 0], y=kmeans.cluster_centers_[:, 1],marker="X",c="red",s=200)
```
